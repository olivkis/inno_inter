**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак marital-status)? Женатыми считаем тех, у кого marital-status начинается с Married (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
high_salary_man = ((data['sex']=='Male') &
                            (data['salary']=='>50K')).sum()

married_man = ((data['sex']=='Male') &
               (data['salary']=='>50K') &
               (data['marital-status'].str.lower().str.startswith('married'))).sum()

all_people=len(data)

married_share = (married_man / all_people) * 100
unmarried_share = ((high_salary_man-married_man) / all_people) * 100

def who_is_better():
  if married_share>unmarried_share:
    return f'Married man earn more on {(married_share-unmarried_share):.2f}%'
  else:
    return f'Unmarried man earn more on {(unmarried_share-married_share):.2f}%'

print(who_is_better())
```

**9. Какое максимальное число часов человек работает в неделю (признак hours-per-week)? Сколько людей работают такое количество часов и каков среди них процент зарабатывающих много?**

```python
max_hours = data['hours-per-week'].max()

max_hour_workers = data.loc[data['hours-per-week'] == max_hours,
                           ['salary', 'hours-per-week']]

max_hour_count = max_hour_workers.shape[0]

rich_share_max_hours = (data.loc[
    data['hours-per-week']==max_hours, 'salary']
    .eq('>50K')
    .mean()*100)

print(
    f"{'Maximum number of hours:':35} {max_hours}\n"
    f"{'Count of max-hour workers:':35} {max_hour_count}\n"
    f"{'Share earning >50K:':35} {rich_share_max_hours:.2f}%"
)
```

**11.Сгруппируйте людей по возрастным группам young, adult, retiree, где:**

*   young соответствует 16-35 лет
*   adult - 35-70 лет
*   retiree - 70-100 лет

**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

```python
def age_group(i):
  if (i>=16) & (i<35):
    return 'young'
  elif (i>=35) & (i<70):
    return 'adult'
  elif (i>=70) & (i<100):
    return 'retiree'

data['AgeGroup']=data['age'].apply(age_group)

col=data.pop('AgeGroup')
data.insert(1, 'AgeGroup', col)
```

```python
high_income = data['salary'].eq('>50K')

max_salary_count = high_income.groupby(data['AgeGroup']).sum()
max_group=max_salary_count.idxmax()
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
occupation = data.groupby('occupation')[['hours-per-week']].agg('count')

def filter_func(group):
    avg_age = group['age'].mean() <= 40
    total_time = (group['hours-per-week'] > 5).all()
    return avg_age and total_time

filtered_groups = data.groupby('occupation').filter(filter_func)

result = (filtered_groups.groupby('occupation')
                         .agg({'age': 'mean',
                               'hours-per-week': 'min'}))
```
