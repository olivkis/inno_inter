**4. Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B**

```python
A = np.random.randint(-3, 3, size=(8, 3))
B = np.random.randint(-2, 2, size=(2, 2))

matching_rows = []

for a, row_a in enumerate(A):
  for b, row_b in enumerate(B):
    if all(set(row_b).issubset(set(row_a)) for row_b in B):
      matching_rows.append(a)
      break

print("A =\n", A)
print("A =\n", B)
print(matching_rows)
```

Задача 4: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.
Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python

x = np.array([2, 2, 2, 3, 3, 3, 5])

# numpy
unique_values, count = np.unique(x, return_counts=True)
print(f'Unique array: {unique_values}\n Count: {count}')

# python
value = []
counts = []

current_value=x[0]
current_count=1

for i in range(1, len(x)):
  if x[i]==current_value:
    current_count +=1
  else:
    value.append(current_value)
    counts.append(current_count)
    current_value = x[i]
    current_count =1

value.append(current_value)
counts.append(current_count)

print(value, counts)
```

Задача 5: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import time
from scipy.spatial.distance import cdist

X = np.array([
    [1, 2],
    [4, 6],
    [7, 8]
])

Y = np.array([
    [2, 5],
    [7, 9],
    [0, 17]
])

# numpy
start_time = time.time()
D=np.zeros((len(X), len(Y)))

for i in range(len(X)):
  for j in range(len(Y)):
    D[i, j]=np.linalg.norm(X[i]-Y[j])

print(f'D = \n{D}\n')
print('Time: ', time.time()-start_time)

# scipy
start = time.time()

D2 = cdist(X, Y, metric='euclidean')
print(f'\nD2 = \n{D2}\n')
print("Time scipy:", time.time() - start)
```
